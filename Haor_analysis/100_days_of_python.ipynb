```python
#The Treasure Island game

print("Welcome to Treasure Island. Our mission is to find the treasure")

direction = input("You are at a crossroad. Left or Right?").lower()

if direction ==  "right":
    print("You got killed by Draco Malfoy. womp womp")
elif direction == "left":
    river = input("You are at a river. Do you want to Swim or Wait for a boat?").lower()
    if river == "swim":
        print("You got attacked by a crocodile. womp womp")
    elif river == "wait":
        door = input("You have 3 doors. Red, Blue and Yellow. choose one").lower()
        if door =="red":
            print("Burnt by fire. womp womp")
        elif door == "blue":
            print("Eaten by beasts. womp womp")
        elif door == "yellow":
            print("You Win")
        else:
            print("Game Over!")
    else:
        print("Game Over")
else: 
    print("Game Over")
```
